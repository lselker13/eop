```python
import pandas as pd
from pathlib import Path
survey_directory_to_country_name_map = {
    'burkina_faso': 'burkina faso',
    'cote_divoire': "cote d'ivoire",
    'south_africa': 'south africa',
    'south_sudan': 'south sudan',
    'yemen': 'yemen, rep.'
}
```

```python

aux_data = pd.read_csv("/data/eop/compiled_country_data/auxiliary_data/auxiliary_data_20251124.csv")
df = aux_data[['country', 'country_code']]
country_name_to_survey_directory_name_map = {v: k for k, v in survey_directory_to_country_name_map.items()}
df['dirname'] = df.country.replace(country_name_to_survey_directory_name_map)

name_map = dict(zip(df["dirname"], df["country_code"]))

base = Path("/data/eop/unfinished_country_data")
dry_run = True

for dir in base.iterdir():
    if not dir.is_dir():
        print(f'skipping non-directory {dir.name}')
        continue
    
    old_name = dir.name
    new_name = name_map.get(old_name, None)
    if new_name is None:
        print(f'Skipping {old_name}: Not in mapping.')
        continue
    if dry_run:
        print(f"Would rename: {old_name} -> {new_name}")
    else:
        dir.rename(dir.parent / new_name)
        print(f"Renamed: {old_name} -> {new_name}")
```
